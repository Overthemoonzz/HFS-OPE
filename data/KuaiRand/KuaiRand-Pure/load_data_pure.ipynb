```python
import pandas as pd
```

```python
df1 = pd.read_csv("data/log_standard_4_08_to_4_21_pure.csv")
df2 = pd.read_csv("data/log_standard_4_22_to_5_08_pure.csv")
```

```python
df = pd.concat([df1, df2], ignore_index=True)
del df1, df2
```

```python
df.columns
```

```python
labels = ['is_click', 'is_like', 'is_follow', 'is_comment', 'is_forward']
labels_p_n = dict()
for label in labels:
    p_df = df[df[label].isin([1])]
    n_df = df[df[label].isin([0])]    
    labels_p_n[label] = (len(p_df), len(n_df), f'{len(p_df)/len(n_df)*100:.4f}')
```

```python
del labels_p_n['is_hate']
labels_p_n
```

```python
user_features = pd.read_csv("data/user_features_pure.csv")

video_features_basic = pd.read_csv("data/video_features_basic_pure.csv")
#video_features_statistics = pd.read_csv("data/video_features_statistic_pure.csv", nrows=10000)
```
